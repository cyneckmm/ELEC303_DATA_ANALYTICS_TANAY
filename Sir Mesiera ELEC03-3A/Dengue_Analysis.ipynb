<p>The internet says about the dengue cases, Dengue is always present in the Philippines due to a combination of factors including its tropical climate, which is favorable for mosquitoes, and the country's high urbanization, which creates abundant breeding sites in urban environments. Other contributing factors include high population density, poor waste management, and insufficient public health infrastructure, all of which allow the Aedes aegypti mosquito vector to thrive and spread the virus effectively. <br>
<br>Favorable conditions
Climate: The Philippines' tropical and subtropical climate provides ideal conditions for mosquitoes to breed and survive year-round. Warmer and wetter years can lead to even more mosquitoes. <br>
<br>Urbanization and population density: Rapid urbanization and high population density in cities like Metro Manila create numerous breeding grounds and facilitate the rapid spread of the virus among densely packed populations. 
Mosquito breeding sites: Aedes aegypti mosquitoes lay eggs in and around water. Water-holding containers like flowerpots, pails, tires, and even scattered garbage that accumulates rainwater are common breeding sites in both urban and rural settings. <br>
<br>Public health and infrastructure issues
Inadequate waste management: Poor solid waste management leads to trash accumulation that can collect rainwater, creating numerous breeding sites for mosquitoes.
Inconsistent prevention practices: Challenges in completely eradicating local breeding sites, often due to a lack of community empowerment and inconsistent preventive practices, allow the mosquito population to persist.
Lack of effective surveillance: An insufficient mosquito surveillance system makes it difficult to effectively track and control mosquito populations and prevent outbreaks. <br>
<br>Globalization
International travel: The country's role as a major transportation hub facilitates the movement of both infected individuals and mosquitoes, further spreading the virus within the country and to other parts of the world. </p>

<h2>Predictives</h2>

```python
dengue['Month'] = dengue['Month'].astype('str')
```

```python
# Combine Month and Year into a Date column
dengue['Date'] = pd.to_datetime(dengue['Month'] + ' ' + dengue['Year'].astype(str))

# Optional: sort by date
dengue = dengue.sort_values('Date')
```

```python
region_df = dengue[dengue['Region'] == 'Region IV-A']
monthly_cases = region_df.groupby('Date')['Dengue_Cases'].sum().reset_index()
dengue
```

<hr>
<h2>Regression</h2>

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Convert dates to ordinal for regression
#monthly_cases['Date_Ordinal'] = monthly_cases['Date'].map(lambda x: x.toordinal())
#other 
monthly_cases['Date_Ordinal'] = monthly_cases['Date'].map(pd.Timestamp.toordinal)

# Reshape for sklearn
X = monthly_cases['Date_Ordinal'].values.reshape(-1, 1) #independent variable 
y = monthly_cases['Dengue_Cases'].values   #dependent variable 

# Fit model
model = LinearRegression()
model.fit(X, y)

# Predict
monthly_cases['Predicted_Cases'] = model.predict(X)

# Evaluate: R-squared
r_squared = model.score(X, y)
print(f"The model explains {r_squared:.1%} of the variation in dengue cases.")
monthly_cases
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(monthly_cases['Date'], monthly_cases['Dengue_Cases'], label='Actual', color='#2E7D32')
plt.plot(monthly_cases['Date'], monthly_cases['Predicted_Cases'], label='Predicted', linestyle='--', color='#81C784')
plt.legend()
plt.title('Linear Regression Forecast of Dengue Cases')
plt.xlabel('Date')
plt.ylabel('Cases')
plt.show()

```
